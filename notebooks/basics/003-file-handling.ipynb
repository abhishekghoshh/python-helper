# Python File handling

## File Handling

The key function for working with files in Python is the `open()` function.

The `open()` function takes two parameters: **filename** and **mode**.

### File Modes

| Mode | Description |
|------|-------------|
| **"r"** | Read — Default. Opens a file for reading; error if the file does not exist. |
| **"a"** | Append — Opens a file for appending; creates the file if it does not exist. |
| **"w"** | Write — Opens a file for writing; creates the file if it does not exist. |
| **"x"** | Create — Creates the file; error if the file already exists. |

### Text or Binary

| Mode | Description |
|------|-------------|
| **"t"** | Text mode — Default. |
| **"b"** | Binary mode — Used for images, videos, etc. |

To open a file for reading it is enough to specify the name of the file:

````python
f = open("out/user.txt")
````

The code above is the same as: </br>

Because "r" for read, and "t" for text are the default values, you do not need to specify them.  </br>
> Note: Make sure the file exists, or else you will get an error.

````python
f = open("out/user.txt", "rt")
````

To open the file, use the built-in `open()` function.

The `open()` function returns a file object, which has a `read()` method for reading the content of the file:

````python
f = open("out/user.txt")
print(f.read())
````

### Using the with statement

You can also use the with statement when opening a file: </br>
Then you do not have to worry about closing your files, the with statement takes care of that. </br>

````python
with open("out/user.txt") as f:
  print(f.read())
````

### Close Files

It is a good practice to always close the file when you are done with it. </br>
If you are not using the with statement, you must write a close statement in order to close the file:  </br>
> Note: You should always close your files. In some cases, due to buffering, changes made to a file may not show until you close the file.

Close the file when you are finished with it:

f = open("out/user.txt")
print(f.readline())
f.close()

### Read Only Parts of the File

By default the `read()` method returns the whole text, but you can also specify how many characters you want to return: </br>
Return the 5 first characters of the file:

````python
with open("out/user.txt") as f:
  print(f.read(5))
````

### Read Lines

You can return one line by using the `readline()` method: </br>
Read one line of the file:

````python
with open("out/user.txt") as f:
  print(f.readline())
````

By calling `readline()` two times, you can read the two first lines:

````python
with open("out/user.txt") as f:
  print("line is :", f.readline())
  print("line is :", f.readline())
````

By looping through the lines of the file, you can read the whole file, line by line:

````python
with open("out/user.txt") as f:
  for line in f:
    print(line)
````

### Write to an Existing File

To write to an existing file, you must add a parameter to the open() function:
- `"a"` - Append - will append to the end of the file
- `"w"` - Write - will overwrite any existing content

````python
with open("out/demofile.txt", "a") as f:
  f.write("Now the file has more content!")

#open and read the file after the appending:
with open("out/demofile.txt") as f:
  print(f.read())
````

### Overwrite Existing Content

To overwrite the existing content to the file, use the `w` parameter:

````python
with open("out/demofile.txt", "w") as f:
  f.write("Woops! I have deleted the content!")

#open and read the file after the overwriting:
with open("out/demofile.txt") as f:
  print(f.read())
````

Create a new file called "myfile.txt": </br>

````python
import os

file_name = "out/myfile.txt"

if os.path.exists(file_name):
    os.remove(file_name)
else:
    with open(file_name, "x") as f:
        f.write("This is a brand new file")
````
