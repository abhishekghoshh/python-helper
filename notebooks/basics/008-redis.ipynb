### Using Redis as a Cache Example

```python
client.set("user:action:save", "Abhishek is saving a person")
```

```python
client.setex("user:1", 60, "Abhishek")  # expires in 60 sec
client.set("user:2", "Nasim", ex = 60)  # expires in 60 sec
```

```python
print(client.get("user:"))
```

```python
user = {"id": 2, "name": "Nasim", "role": "engineer"}

client.set("user:2", json.dumps(user))
```

```python
client.get("user:1")
```

```python
users = []

for key in client.scan_iter("user:*"):
    data = client.get(key)
    users.append(json.loads(data))

print(users)
```

### Storing hash instead of json

```python
person = {
    "id": "1",
    "name": "Abhishek",
    "role": "engineer"
}
# Save person as a Redis Hash
client.hset("person:1", mapping=person)

# Retrieve full person
data = client.hgetall("person:1")
print(data)              # {'id': '1', 'name': 'Abhishek', 'role': 'engineer'}

# Retrieve single field
print(data["name"])      # Abhishek
```

```python
people = []

for key in client.scan_iter("person:*"):
    people.append(client.hgetall(key))

print(people)
```

```python
client.zadd("top_score", {"abhishek Ghosh" : 1})
client.zadd("top_score", {"Nasim Molla" : 2})
client.zadd("top_score", {"Pranay Mandal" : 3})
client.zadd("top_score", {"Abhishek Saha" : 4})

job_count = client.zcard("top_score")
print("job count is", job_count)

resp = client.zrange("top_score", 0, 2, desc=False, withscores=True)
print(resp)

resp = client.zrange("top_score", 0, 100, desc=True, withscores=True)
print(resp)

resp = client.zrange("top_score", 0, -1, desc=True, withscores=True)
print(resp)


job_count = client.zcard("top_")
print("job count is", job_count)

resp = client.zrange("top_", 0, 5, desc=True, withscores=True)
print(resp)
```
