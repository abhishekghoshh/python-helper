```python
import redis
import json


try:
    client = redis.Redis(
        host="redis",
        port=6379,
        db=0,
        decode_responses=True
    )

    client.ping()
    print("Connected to Redis!")
    client.close()
except redis.ConnectionError:
    print("Failed to connect to Redis!")
```

```python
client.set("name", "Abhishek Ghosh")
```

```python
client.get("name")
```

### Using Redis as a Cache Example

```python
client.setex("user:1", 60, "Abhishek")  # expires in 60 sec
```

```python
print(client.get("user:"))
```

```python
user = {"id": 2, "name": "Nasim", "role": "engineer"}

client.set("user:2", json.dumps(user))
```

```python
client.get("user:1")
```

```python
users = []

for key in client.scan_iter("user:*"):
    data = client.get(key)
    users.append(json.loads(data))

print(users)
```

### Storing hash instead of json

```python
person = {
    "id": "1",
    "name": "Abhishek",
    "role": "engineer"
}
# Save person as a Redis Hash
client.hset("person:1", mapping=person)

# Retrieve full person
data = client.hgetall("person:1")
print(data)              # {'id': '1', 'name': 'Abhishek', 'role': 'engineer'}

# Retrieve single field
print(data["name"])      # Abhishek
```

```python
people = []

for key in client.scan_iter("person:*"):
    people.append(client.hgetall(key))

print(people)
```
