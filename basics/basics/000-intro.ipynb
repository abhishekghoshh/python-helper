```python
import sys

print(f"Python version: {sys.version}")
print(f"Platform: {sys.platform}")
print(f"Executable path: {sys.executable}")
print(f"Byte order: {sys.byteorder}")
print(f"Copyright: {sys.copyright}")
```

```python
import platform

print(f"System: {platform.system()}")
print(f"Node Name: {platform.node()}")
print(f"Release: {platform.release()}")
print(f"Version: {platform.version()}")
print(f"Machine: {platform.machine()}")
print(f"Processor: {platform.processor()}")
print(f"Platform: {platform.platform()}")
```

Semicolons are optional in Python. </br>
You can write multiple statements on one line by separating them with ; but this is rarely used because it makes it hard to read:

```python
print("Hello"); print("How are you?"); print("Bye bye!")
```

However, if you put two statements on the same line without a separator (newline or ;), Python will give an error:

```python
print("Python is fun!") print("Really!")
```

```python
print("2 + 2 is : "+ str((2+2)))

num1 = num2 = 2
result = num1 + num2
print(f"{num1} + {num2} is : {result}")

print("{} + {} is : {}".format(num1, num2, result))
```

Text in Python must be inside quotes. You can use either " double quotes or ' single quotes: </br>
If you forget to put the text inside quotes, Python will give an error:

```python
print("This will work!")
print('This will also work!')
```

By default, the print() function ends with a new line. </br>
If you want to print multiple words on the same line, you can use the end parameter: </br>
Note that we add a space after end="  " for better readability.

```python
print("Hello World!", end="  ")
print("I will print on the same line.")
```
